```python
from keras.models import load_model
import pickle
import pandas as pd
import tensorflow as tf
import numpy as np
```

```python
with open('label_encoder.pkl', 'rb') as f:
    label_encoder = pickle.load(f)
    
```

```python
with open('onehot_encoder.pkl', 'rb') as f:
    onehot_encoder = pickle.load(f)
```

```python
with open('scaler.pkl', 'rb') as f:
    scaler = pickle.load(f)
```

```python
model = load_model('titanic_model.keras')
```

```python
# Pclass	Sex	SibSp	Parch	Fare	Embarked_Cherbourg	Embarked_Queenstown	Embarked_Southampton
```

```python
# user = {'Pclass': 3,'Sex':'female','SibSp': 1,'Parch': 0,'Fare': 11,'Embarked':'Queenstown'}
```

```python
user = {'Pclass': 1,'Sex':'male','SibSp': 2,'Parch': 0,'Fare': 76,'Embarked':'Cherbourg'}
```

```python
user = pd.DataFrame([user])
```

```python
user
```

```python
user['Sex'] = label_encoder.transform(user['Sex'])
```

```python
user
```

```python
embarked = onehot_encoder.transform(user[['Embarked']])
```

```python
embarked = pd.DataFrame(embarked, columns=onehot_encoder.get_feature_names_out())
```

```python
embarked
```

```python
data = pd.concat([user.drop('Embarked', axis=1), embarked], axis=1)
```

```python
data
```

```python
data[['Pclass', 'SibSp', 'Parch', 'Fare']] = scaler.transform(data[['Pclass', 'SibSp', 'Parch', 'Fare']])
```

```python
data
```

```python
y = model.predict(data)
```

```python
y = y[0][0]
```

### Probability of Survival
Chance of survival of passenger is 0.65
- if model > 0.5 then --> Survived
- else --> Not Survived

```python
if y > 0.5:
    print(f"Chance of survival of passenger is {y:.2f}")
    print("Passenger is likely to Survive")
else:
    print(f"Chance of survival of passenger is {y:.2f}")
    print("Passenger is likely to Not Survive") 
```
