### Feature Scaling :
1. Standardization : It is a technique that transforms the data to have a mean of 0 and a standard deviation of 1. It is also known as Z-score normalization. It is useful when the data has a normal distribution and when the features have different scales. It is less sensitive to outliers than Min-Max Scaling.
2. Min-Max Scaling : It is a technique that transforms the data to a fixed range, usually [0, 1]. It is useful when the data does not have a normal distribution and when the features have different scales. It is sensitive to outliers, as it uses the minimum and maximum values of the data to scale it.

### Sequential Model : ANN Implementation
1. Sequential Network
2. Dense Layer : It is hidden layer in the ANN. In this we mention the number of neurons in the hidden layer and the activation function to be used in that layer. The number of neurons in the hidden layer is a hyperparameter that we need to tune. The activation function is also a hyperparameter that we need to tune.
3. Activation Function : It is a function that is applied to the output of each neuron in the hidden layer. It is used to introduce non-linearity in the model. It helps the model to learn complex patterns in the data. Some common activation functions are ReLU, Sigmoid, Tanh, etc. The activation function should be sigmoid for the output layer because it is a binary classification problem. If it is a multi-class classification problem, we will use softmax activation function in the output layer. If it is a regression problem, we will not use any activation function in the output layer.
4. Optimizer : It is an algorithm that is used to update the weights of the neurons in the hidden layer during the training process. It is used to minimize the loss function. Some common optimizers are SGD, AdaGrad,Adadelta, Adam, RMSprop, etc.
5. Loss function : It is a function that is used to measure the error between the predicted output and the actual output. It is used to train the model. Some common loss functions are Mean Squared Error, Mean Absolute Error, Binary Crossentropy, Categorical Crossentropy, etc. It is a classification problem, we will use Binary Crossentropy loss function. If it is a multi-class classification problem, we will use Categorical Crossentropy loss function. If it is a regression problem, we will use Mean Squared Error loss function.
6. Metrics : It is a function that is used to evaluate the performance of the model. Some common metrics are Accuracy, Precision, Recall, F1 Score, etc. It is a classification problem, we will use Accuracy as the metric. If it is a multi-class classification problem, we will use Accuracy as the metric. If it is a regression problem, we will use Mean Absolute Error as the metric.
7. TensorBoard : It is a tool that is used to visualize the training process of the model. It is used to monitor the loss and accuracy of the model during the training process. It is also used to visualize the weights of the neurons in the hidden layer.
4. Output Layer : It is the last layer in the ANN. It is used to make predictions. The number of neurons in the output layer depends on the type of problem we are trying to solve. For example, if we are trying to solve a binary classification problem, we will have 1 neuron in the output layer. If we are trying to solve a multi-class classification problem, we will have as many neurons in the output layer as the number of classes. If we are trying to solve a regression problem, we will have 1 neuron in the output layer.

```python
import tensorflow as tf
```

### Model Building :

```python
import tensorflow as tf
from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense
from keras.callbacks import EarlyStopping, TensorBoard

print('TensorFlow:', tf.__version__)
```

```python
(X_train.shape[1],)
## In sequential Layer always give the input shape of the 
# data in the first layer only. 
# After that, it will automatically 
# understand the shape of the data.
## As list as given above.
```

```python
model = Sequential([Dense(64, activation='relu', input_shape=(X_train.shape[1],)), ## First layer with input shape
            Dense(32, activation='relu'), ## Second layer
            Dense(1, activation='sigmoid')]) ## Output layer with sigmoid activation for binary classification
```

```python
model.summary()
```

Parameters :
- First layer : 64 neurons --> 64 * (number of input features) + 64 (bias) = 64 * (number of input features) + 64 = 64 * x  +  64  = 576
- Second layer : 32 neurons --> 32 * 64 + 32 (bias) = 32 * 64 + 32 = 2080
- Output layer : 1 neuron --> 1 * 32 + 1 (bias) = 33


### Model Compilation : --> Optimizer, Loss function, Metrics 
It is the process of configuring the model for training. It is done before training the model. It is used to specify the optimizer, loss function, and metrics to be used during the training process. The optimizer is used to update the weights of the neurons in the hidden layer during the training process. The loss function is used to measure the error between the predicted output and the actual output. The metrics are used to evaluate the performance of the model.

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
### adam --> learning rate = 0.001 (default) --> It is the step size at which the model learns. It is used to update the weights of the neurons in the hidden layer during the training process. 
# The learning rate is a hyperparameter that can be tuned to improve the performance of the model.
# A higher learning rate can lead to faster 
# convergence but may also lead to overshooting 
# the optimal weights, while a lower learning rate 
# can lead to slower convergence but may also lead to 
# better performance.
```

If want to create the optimizer by your own then we can use the following code :

```python
opt = keras.optimizers.Adam(learning_rate=0.01)
```

```python
model.compile(optimizer=opt, loss='binary_crossentropy', metrics=['accuracy'])
```

```python
stopping_callback = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
```

```python
model.fit(X_train,y_train, validation_data=(X_valid, y_valid), epochs=50, callbacks=[stopping_callback])
```
